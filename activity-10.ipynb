```python
import os
from groq import Groq
from dotenv import load_dotenv

load_dotenv()


os.environ["GROQ_API_KEY"]

Client = Groq(api_key=os.environ["GROQ_API_KEY"])
```

```python
#AI/ML Tutor Instructions 
AI_ML_TUTOR_PROMPT_SYSTEM = """
You are an expert AI/ML learning tutor with deep knowledge in:

- Machine Learning fundamentals
- Supervised, Unsupervised, Reinforcement Learning
- Deep Learning and Neural Networks
- Natural Language Processing (NLP)
- Computer Vision
- Large Language Models (LLMs)
- Python for AI/ML
- Statistics and Mathematics for ML
- Data preprocessing and feature engineering
- Model evaluation and optimization

Your teaching approach:

1. Explain complex concepts in simple, progressive terms.
2. Use real-world examples and analogies.
3. Provide code examples when relevant.
4. Break difficult problems into manageable steps.
5. Ask clarifying questions when necessary.
6. Provide practical examples and challenges.
7. Be patient and student-friendly.
"""

conversation_history = [
    {
        "role": "system",
        "content": AI_ML_TUTOR_PROMPT_SYSTEM

    }
]

#CHATBOT FUNCTION 

def Chatbot(user_message):
    conversation_history.append({
        "role": "user",
        "content": user_message

    })

    response = Client.chat.completions.create(
        model = "openai/gpt-oss-120b",
        messages = conversation_history
    )
    
    answer = response.choices[0].message.content


    conversation_history.append({
        "role": "assistant",
        "content": answer 
    })
    return answer 

    #chat loop

    while true:
        user_input = input("You: ")
        if user_input.lower() == "exit":
            print("Tutor; bye!")
            break

        answer = chatbot(user_input)
        print("Tutor:", answer)





```
